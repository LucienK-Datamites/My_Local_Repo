```python
from typing import List
import torch
import matplotlib.pyplot as plt
from diffusers import StableDiffusionPipeline, DDPMScheduler
```

```python
def load_model (model_id: str)-> StableDiffusionPipeline :
  """Load model with provided mode_id."""
  return StableDiffusionPipeline.from_pretrained(
      model_id,
      torch_dtype=torch.float16,
      variant="fp16",
      use_auth_token=True
  ).to ("cuda")

def generate_images (
    pipe:StableDiffusionPipeline,
    prompts:List[str]
)-> torch.tensor:
    """Generate images based on provided prompts."""
    with torch.autocast ("cuda"):
      images = pipe (prompts).images
      return images

def render_images (images: torch.Tensor):
  """plot the generate images."""
  plt.figure(figsize=(25, 25))
  for i, img in enumerate (images):
      plt.subplot(1,2,i+1)
      plt.imshow(img)
      plt.axis("off")
  plt.show()
```

```python
# Lets Execute now
model_id="CompVis/stable-diffusion-v1-4"
prompts=["A hyper-realistic photo of a friendly lion",
         "A stylized oil paining of NYC Brownstone"
         ]
pipe = load_model (model_id)
images=generate_images(pipe, prompts)
render_images(images)
```
